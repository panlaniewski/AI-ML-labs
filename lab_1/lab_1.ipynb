## Предобработка данных. Знакомство с pandas

Импортируем необходимые библиотеки:

```python
import pandas as pd
```

Загружаем набор данных:

```python
df = pd.read_csv("data/StudentsPerformance_lab.csv")
```

Выводим данные из датасета:

```python
df
```

Выводим количество пропусков в каждой колонке:

```python
df.isna().sum()
```

Видим, что значения пропущены в колонках **"parental level of education", "reading score", "writing score"**. Чтобы заполнить пропущенные значения, посчитаем среднее, медиану и моду:

```python
reading_score_mean = df["reading score"].mean()
reading_score_mean
```

```python
parental_level_mode = df["parental level of education"].mode()[0]
parental_level_mode
```

```python
writing_score_median = df["writing score"].median()
writing_score_median
```

Заполняем пропуски посчитанными значениями:

```python
df["reading score"] = df["reading score"].fillna(reading_score_mean)
df["parental level of education"] = df["parental level of education"].fillna(parental_level_mode)
df["writing score"] = df["writing score"].fillna(writing_score_median)
```

Проверяем, что пропусков действительно больше нет:

```python
df.isna().sum()
```

Проведём нормализацию числовых данных:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
df[['math score', 'reading score', 'writing score']] = scaler.fit_transform(df[['math score', 'reading score', 'writing score']])
```

Выберем колонки с категориальными признаками и произведём кодирование в численный вид:

```python
cat_cols = df.select_dtypes(include=["str"]).columns
cat_cols
```

```python
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

Выведем конечный вид предобработанного датасета:

```python
df
```

Сохраним датасет:

```python
df.to_csv("data/StudentsPerformance_processed.csv")
```
