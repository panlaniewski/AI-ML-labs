## Предобработка данных. Знакомство с pandas

Импортируем необходимые библиотеки:

```python
import pandas as pd
```

Загружаем набор данных:

```python
df = pd.read_csv("data/bigmart_data_lab.csv")
```

Выводим данные из датасета:

```python
df
```

Выводим количество пропусков в каждой колонке:

```python
df.isna().sum()
```

Видим, что значения пропущены в колонках **Item_Weight, Item_Type, Item_MRP, Outlet_Size**. Чтобы заполнить пропущенные значения, посчитаем среднее, медиану и моду:

```python
item_weight_mean = df["Item_Weight"].mean()
item_weight_mean
```

```python
outlet_size_mode = df["Outlet_Size"].mode()[0]
outlet_size_mode
```

```python
item_type_mode = df["Item_Type"].mode()[0]
item_type_mode
```

```python
item_mrp_median = df["Item_MRP"].median()
item_mrp_median
```

Заполняем пропуски посчитанными значениями:

```python
df["Item_Weight"] = df["Item_Weight"].fillna(item_weight_mean)
df["Outlet_Size"] = df["Outlet_Size"].fillna(outlet_size_mode)
df["Item_Type"] = df["Item_Type"].fillna(item_type_mode)
df["Item_MRP"] = df["Item_MRP"].fillna(item_mrp_median)
```

Проверяем, что пропусков действительно больше нет:

```python
df.isna().sum()
```

Проведём нормализацию числовых данных:

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
df[['Item_Weight', 'Item_MRP', 'Item_Visibility']] = scaler.fit_transform(df[['Item_Weight', 'Item_MRP', 'Item_Visibility']])
```

```python
df.nunique()
```

Выберем колонки с категориальными признаками и произведём кодирование в численный вид:

```python
cat_cols = ["Item_Type", "Item_Fat_Content", "Outlet_Size", "Outlet_Location_Type", "Outlet_Type"]
cat_cols
```

```python
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)
```

Выведем конечный вид предобработанного датасета:

```python
df
```

```python
df.drop(columns=["Item_Identifier", "Outlet_Identifier"], inplace=True)
df
```

Сохраним датасет:

```python
df.to_csv("data/bigmart_data_processed.csv")
```
